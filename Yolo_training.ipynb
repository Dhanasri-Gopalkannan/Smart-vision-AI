```python
"""
COMPLETE YOLOv8 OBJECT DETECTION TRAINING
All steps in one file - Ready to run in Colab
"""

# ========== STEP 1: INSTALLATION ==========
print("="*70)
print("STEP 1: INSTALLING YOLOv8")
print("="*70)

!pip install ultralytics -q
!pip install matplotlib seaborn -q

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import shutil
import random
import time
from pathlib import Path
from IPython.display import Image, display
import warnings
warnings.filterwarnings('ignore')

from ultralytics import YOLO

print(" YOLOv8 installed successfully!")

# Check GPU
import torch
print(f"\n Hardware:")
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

# ========== STEP 2: DATASET PATH ==========
print("\n" + "="*70)
print("STEP 2: DATASET SETUP")
print("="*70)

# Your dataset path - UPDATE THIS IF NEEDED
dataset_path = "/content/dataset_processing_output/detection_processed"

if not os.path.exists(dataset_path):
    print(f" Dataset not found at: {dataset_path}")
    print("\n Searching for dataset...")

    # Try alternative paths
    alternatives = [
        "/content/dataset_processing_output/detection_processed",
        "/content/detection_processed",
        "/content/yolo_dataset",
        "/content/dataset"
    ]

    for alt in alternatives:
        if os.path.exists(alt):
            dataset_path = alt
            print(f" Found dataset at: {dataset_path}")
            break
    else:
        print(" No dataset found!")
        print("\n Contents of /content:")
        !ls -la /content/
        exit()

print(f"\n Using dataset: {dataset_path}")

# ========== STEP 3: CHECK DATASET STRUCTURE ==========
print("\n Checking dataset structure...")

# Check train set
train_images = os.path.join(dataset_path, 'train', 'images')
train_labels = os.path.join(dataset_path, 'train', 'labels')

if os.path.exists(train_images):
    train_img_count = len([f for f in os.listdir(train_images) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
    print(f" train/images: {train_img_count} images")
else:
    print(f" train/images not found")
    train_img_count = 0

if os.path.exists(train_labels):
    train_lbl_count = len([f for f in os.listdir(train_labels) if f.endswith('.txt')])
    print(f" train/labels: {train_lbl_count} label files")
else:
    print(f" train/labels not found")
    train_lbl_count = 0

# Check val set
val_images = os.path.join(dataset_path, 'val', 'images')
val_labels = os.path.join(dataset_path, 'val', 'labels')

if os.path.exists(val_images):
    val_img_count = len([f for f in os.listdir(val_images) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
    print(f" val/images: {val_img_count} images")
else:
    print(f" val/images not found")
    val_img_count = 0

if os.path.exists(val_labels):
    val_lbl_count = len([f for f in os.listdir(val_labels) if f.endswith('.txt')])
    print(f" val/labels: {val_lbl_count} label files")
else:
    print(f" val/labels not found")
    val_lbl_count = 0

# ========== STEP 4: FIX DATA.YAML ==========
print("\n Creating/fixing data.yaml...")

# Class names (26 classes from your dataset)
class_names = [
    'person', 'bicycle', 'car', 'motorcycle', 'airplane', 'bus', 'train', 'truck',
    'traffic light', 'stop sign', 'bench', 'bird', 'cat', 'dog', 'horse', 'cow',
    'elephant', 'bottle', 'cup', 'bowl', 'pizza', 'cake', 'chair', 'couch',
    'potted plant', 'bed'
]

# Create data.yaml
data_yaml = {
    'path': dataset_path,
    'train': 'train/images',
    'val': 'val/images',
    'test': 'val/images',  # Use val as test if test not available
    'nc': 26,
    'names': class_names
}

data_yaml_path = os.path.join(dataset_path, 'data.yaml')
with open(data_yaml_path, 'w') as f:
    yaml.dump(data_yaml, f, default_flow_style=False)

print(f" Created: {data_yaml_path}")
print("\n data.yaml contents:")
print(yaml.dump(data_yaml, default_flow_style=False))

# ========== STEP 5: CHECK LABEL FORMAT ==========
print("\n Checking label format...")

# Check a sample label file
if train_lbl_count > 0:
    label_files = os.listdir(train_labels)
    sample_label = os.path.join(train_labels, label_files[0])

    with open(sample_label, 'r') as f:
        lines = f.readlines()
        print(f"\n Sample label file: {label_files[0]}")
        print(f"   Contains {len(lines)} objects")

        for i, line in enumerate(lines[:3]):  # Show first 3 objects
            parts = line.strip().split()
            if len(parts) == 5:
                class_id = int(float(parts[0]))
                x, y, w, h = map(float, parts[1:5])
                class_name = class_names[class_id] if class_id < 26 else f"class_{class_id}"

                # Check if normalized (should be between 0-1)
                if 0 <= x <= 1 and 0 <= y <= 1 and 0 <= w <= 1 and 0 <= h <= 1:
                    status = " Normalized"
                else:
                    status = " NOT normalized (values > 1)"

                print(f"   Object {i+1}: Class {class_id} ({class_name})")
                print(f"     Coordinates: x={x:.4f}, y={y:.4f}, w={w:.4f}, h={h:.4f}")
                print(f"     Status: {status}")
            else:
                print(f"    Invalid format: {line.strip()}")

# ========== STEP 6: TRAINING CONFIGURATION ==========
print("\n" + "="*70)
print("STEP 6: TRAINING CONFIGURATION")
print("="*70)


model_size = 's'  # Using small for good balance

# Training parameters
epochs = 50
batch_size = 16
img_size = 640

print(f"\n Training Configuration:")
print(f"   Model: YOLOv8{model_size.upper()}")
print(f"   Epochs: {epochs}")
print(f"   Batch size: {batch_size}")
print(f"   Image size: {img_size}")
print(f"   Classes: 26")
print(f"   Training images: {train_img_count}")
print(f"   Validation images: {val_img_count}")

# ========== STEP 7: LOAD PRETRAINED MODEL ==========
print("\n Loading pre-trained YOLOv8...")

model = YOLO(f'yolov8{model_size}.pt')
print(f" Loaded YOLOv8{model_size.upper()}")

# ========== STEP 8: START TRAINING ==========
print("\n" + "="*70)
print("STEP 8: YOLO TRAINING")
print("="*70)
print(" Training started - this will take 30-60 minutes...")
print("   Each epoch takes ~7-10 seconds")
print("   Total time: ~50 epochs × 8 seconds = ~7 minutes\n")

# Create results folder
results_dir = "/content/yolo_training_results"
os.makedirs(results_dir, exist_ok=True)

# Start training
results = model.train(
    data=data_yaml_path,
    epochs=epochs,
    imgsz=img_size,
    batch=batch_size,
    workers=4,
    device=0,
    patience=15,
    save=True,
    save_period=5,
    project=results_dir,
    name='yolov8_training',
    exist_ok=True,
    pretrained=True,
    optimizer='auto',
    verbose=True,
    plots=True
)

print("\n Training completed!")

# ========== STEP 9: EVALUATION ==========
print("\n" + "="*70)
print("STEP 9: MODEL EVALUATION")
print("="*70)

# Load best model
best_model_path = os.path.join(results_dir, 'yolov8_training', 'weights', 'best.pt')
if os.path.exists(best_model_path):
    model = YOLO(best_model_path)
    print(f" Loaded best model")
else:
    model = YOLO(os.path.join(results_dir, 'yolov8_training', 'weights', 'last.pt'))
    print(" Loaded last model")

# Run evaluation
print("\n Evaluating on validation set...")
metrics = model.val(
    data=data_yaml_path,
    imgsz=img_size,
    batch=batch_size,
    conf=0.25,
    iou=0.6,
    device=0,
    plots=True
)

# Display metrics
print("\n DETECTION METRICS:")
print(f"   mAP@0.5: {metrics.box.map50:.4f}")
print(f"   mAP@0.5:0.95: {metrics.box.map:.4f}")
print(f"   Precision: {metrics.box.mp:.4f}")  # Mean precision
print(f"   Recall: {metrics.box.mr:.4f}")     # Mean recall

# Compare with expected
print(f"\n EXPECTED vs ACTUAL:")
print(f"   Expected mAP@0.5: > 0.85")
print(f"   Actual mAP@0.5: {metrics.box.map50:.4f}")
if metrics.box.map50 >= 0.85:
    print("    Target achieved!")
else:
    print("    Below target - may need more epochs or data augmentation")

# ========== STEP 10: INFERENCE SPEED TEST ==========
print("\n Testing inference speed...")

# Create dummy image
dummy_img = np.random.randint(0, 255, (640, 640, 3), dtype=np.uint8)

# Warmup
for _ in range(10):
    _ = model(dummy_img, verbose=False)

# Speed test
times = []
for _ in range(50):
    start = time.time()
    _ = model(dummy_img, verbose=False)
    times.append(time.time() - start)

avg_time = np.mean(times) * 1000  # Convert to ms
fps = 1000 / avg_time

print(f"\n Inference Speed:")
print(f"   Average time: {avg_time:.2f} ms per image")
print(f"   FPS: {fps:.2f}")
print(f"   Expected FPS: 30-50")
if 30 <= fps <= 50:
    print("    Speed target achieved!")
else:
    print("    Speed target not met")

# ========== STEP 11: VISUALIZE PREDICTIONS ==========
print("\n Visualizing predictions...")

# Find validation images
val_images_path = os.path.join(dataset_path, 'val', 'images')
if os.path.exists(val_images_path):
    val_images = [f for f in os.listdir(val_images_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

    if val_images:
        # Take 3 random images
        sample_images = random.sample(val_images, min(3, len(val_images)))

        for img_name in sample_images:
            img_path = os.path.join(val_images_path, img_name)

            # Run prediction
            results = model.predict(
                img_path,
                conf=0.25,
                save=True,
                project=results_dir,
                name='predictions'
            )

            # Display result
            pred_path = os.path.join(results_dir, 'predictions', img_name)
            if os.path.exists(pred_path):
                print(f"\n📷 Image: {img_name}")
                display(Image(filename=pred_path, width=600))

                # Show detections
                for r in results:
                    if len(r.boxes) > 0:
                        print(f"   Detected {len(r.boxes)} objects:")
                        for i, box in enumerate(r.boxes[:3]):
                            cls_id = int(box.cls[0])
                            conf = float(box.conf[0])
                            class_name = class_names[cls_id] if cls_id < 26 else f"class_{cls_id}"
                            print(f"     {i+1}. {class_name}: {conf:.2f}")

# ========== STEP 12: SAVE ALL RESULTS ==========
print("\n" + "="*70)
print("STEP 12: SAVING RESULTS")
print("="*70)

# Create final results folder
final_dir = "/content/yolo_final_complete"
os.makedirs(final_dir, exist_ok=True)

# Copy important files
files_to_save = [
    (best_model_path, "best_model.pt"),
    (os.path.join(results_dir, 'yolov8_training', 'weights', 'last.pt'), "last_model.pt"),
    (data_yaml_path, "dataset.yaml"),
    (os.path.join(results_dir, 'yolov8_training', 'results.csv'), "training_results.csv"),
    (os.path.join(results_dir, 'yolov8_training', 'args.yaml'), "training_args.yaml"),
]

print("\n Saving files:")
for src, dst in files_to_save:
    if os.path.exists(src):
        shutil.copy2(src, os.path.join(final_dir, dst))
        print(f"    {dst}")

# Copy plots
plots_dir = os.path.join(results_dir, 'yolov8_training')
if os.path.exists(plots_dir):
    for file in os.listdir(plots_dir):
        if file.endswith('.png'):
            shutil.copy2(os.path.join(plots_dir, file), final_dir)
            print(f"    Plot: {file}")

# ========== STEP 13: CREATE INFERENCE SCRIPT ==========
print("\n Creating inference script for VS Code...")

inference_code = '''"""
YOLOv8 Inference Script
Use this in VS Code with your trained model
"""

from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Class names (26 classes)
class_names = [
    'person', 'bicycle', 'car', 'motorcycle', 'airplane', 'bus', 'train', 'truck',
    'traffic light', 'stop sign', 'bench', 'bird', 'cat', 'dog', 'horse', 'cow',
    'elephant', 'bottle', 'cup', 'bowl', 'pizza', 'cake', 'chair', 'couch',
    'potted plant', 'bed'
]

# Load your trained model
model = YOLO('best_model.pt')

def detect_objects(image_path, confidence=0.25):
    """
    Detect objects in an image
    """
    print(f"\\n Processing: {image_path}")

    # Run detection
    results = model(image_path, conf=confidence)

    # Show results
    for r in results:
        # Plot image with boxes
        im_array = r.plot()
        im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB)

        plt.figure(figsize=(12, 8))
        plt.imshow(im_rgb)
        plt.axis('off')
        plt.title(f'Detection Results (Confidence: {confidence})')
        plt.show()

        # Print detections
        print(f"\\n Found {len(r.boxes)} objects:")
        for i, box in enumerate(r.boxes):
            cls_id = int(box.cls[0])
            confidence = float(box.conf[0])
            class_name = class_names[cls_id] if cls_id < len(class_names) else f"class_{cls_id}"

            # Get bounding box coordinates
            x1, y1, x2, y2 = box.xyxy[0].tolist()

            print(f"  {i+1}. {class_name}: {confidence:.2f}  [{int(x1)}, {int(y1)}, {int(x2)}, {int(y2)}]")

    return results

# Example usage
if __name__ == "__main__":
    # Test on an image
    detect_objects("test_image.jpg", confidence=0.25)
'''

with open(os.path.join(final_dir, "inference.py"), "w") as f:
    f.write(inference_code)

print(" Created inference.py")

# ========== STEP 14: CREATE REQUIREMENTS ==========
with open(os.path.join(final_dir, "requirements.txt"), "w") as f:
    f.write("ultralytics\nopencv-python\nmatplotlib")

print(" Created requirements.txt")

# ========== STEP 15: CREATE SUMMARY REPORT ==========
print("\n Creating summary report...")

report = f"""
YOLOv8 OBJECT DETECTION TRAINING REPORT
{'='*60}

 DATASET INFORMATION:
- Training images: {train_img_count}
- Validation images: {val_img_count}
- Classes: 26
- Class names: {class_names}

 TRAINING CONFIGURATION:
- Model: YOLOv8{model_size.upper()}
- Epochs: {epochs}
- Image size: {img_size}
- Batch size: {batch_size}
- Optimizer: Auto (AdamW)

 PERFORMANCE METRICS:
- mAP@0.5: {metrics.box.map50:.4f}
- mAP@0.5:0.95: {metrics.box.map:.4f}
- Precision: {metrics.box.mp:.4f}
- Recall: {metrics.box.mr:.4f}

 INFERENCE SPEED:
- Average time: {avg_time:.2f} ms per image
- FPS: {fps:.2f}

 EXPECTED vs ACTUAL:
- Expected mAP@0.5 > 0.85: {' ACHIEVED' if metrics.box.map50 >= 0.85 else ' NOT MET'}
- Expected FPS 30-50: {' ACHIEVED' if 30 <= fps <= 50 else ' NOT MET'}

 FILES SAVED:
- best_model.pt - Best trained model (epoch {metrics.epoch if hasattr(metrics, 'epoch') else '?'})
- last_model.pt - Last epoch model
- dataset.yaml - Dataset configuration
- training_results.csv - Training history
- *.png - Training plots and visualizations
- inference.py - Ready-to-use inference script
- requirements.txt - Python dependencies

{'='*60}
"""

with open(os.path.join(final_dir, "report.txt"), "w") as f:
    f.write(report)

print(report)

# ========== STEP 16: CREATE ZIP FOR DOWNLOAD ==========
print("\n" + "="*70)
print("STEP 16: CREATING DOWNLOAD PACKAGE")
print("="*70)

import zipfile

zip_path = "/content/yolov8_complete_results.zip"

with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for root, dirs, files in os.walk(final_dir):
        for file in files:
            file_path = os.path.join(root, file)
            arcname = os.path.relpath(file_path, final_dir)
            zipf.write(file_path, arcname)

zip_size = os.path.getsize(zip_path) / (1024 * 1024)
print(f" Zip created: {zip_path}")
print(f" Size: {zip_size:.2f} MB")

# ========== STEP 17: DOWNLOAD ==========
print("\n" + "="*70)
print("STEP 17: DOWNLOAD RESULTS")
print("="*70)

from google.colab import files

print("\n Downloading to your computer...")
files.download(zip_path)

# ========== STEP 18: SAVE TO GOOGLE DRIVE (OPTIONAL) ==========
print("\n" + "="*70)
print("STEP 18: SAVE TO GOOGLE DRIVE (Optional)")
print("="*70)

save_to_drive = input("\n Also save to Google Drive? (y/n): ").strip().lower()
if save_to_drive == 'y':
    from google.colab import drive
    drive.mount('/content/drive')

    drive_path = "/content/drive/MyDrive/YOLO_Training_Results"
    os.makedirs(drive_path, exist_ok=True)

    shutil.copy2(zip_path, drive_path)
    shutil.copytree(final_dir, os.path.join(drive_path, "results"), dirs_exist_ok=True)

    print(f" Saved to Google Drive: {drive_path}")

print("\n" + "="*70)
print(" YOLOv8 TRAINING COMPLETED SUCCESSFULLY!")
print("="*70)
print(f"\n Results saved in: {final_dir}")
print(f" Zip file: {zip_path}")
print(f" Download complete - check your browser downloads!")
print("\n NEXT STEPS:")
print("   1. Extract the zip file on your computer")
print("   2. Open in VS Code")
print("   3. Install requirements: pip install -r requirements.txt")
print("   4. Run inference: python inference.py")
print("\n All done!")
```
