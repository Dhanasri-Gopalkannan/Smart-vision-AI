```python
"""
ULTRA SIMPLE TRANSFER LEARNING - READY TO RUN
Your dataset is already at: /content/dataset_processing_output
"""

# ========== STEP 1: INSTALL ==========
!pip install tensorflow pandas matplotlib -q

# ========== STEP 2: CHECK DATASET ==========
import os

print("🔍 Checking your dataset...")
dataset_path = '/content/dataset_processing_output'

if os.path.exists(dataset_path):
    print(f"✅ Dataset found at: {dataset_path}")

    # List contents
    items = os.listdir(dataset_path)
    print(f"\n📁 Contents: {items}")

    # Check for train folder
    if 'train' in items:
        train_path = os.path.join(dataset_path, 'train')
        print(f"\n✅ Found train folder: {train_path}")

        # Check train contents
        train_classes = os.listdir(train_path)
        print(f"📊 Train has {len(train_classes)} classes")
        print(f"Classes: {train_classes[:10]}{'...' if len(train_classes) > 10 else ''}")

        # Count images in first class
        if train_classes:
            first_class = os.path.join(train_path, train_classes[0])
            images = [f for f in os.listdir(first_class) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
            print(f"📷 Sample: '{train_classes[0]}' has {len(images)} images")
    else:
        print("❌ No 'train' folder found in dataset")
        exit()
else:
    print(f"❌ Dataset not found at: {dataset_path}")
    exit()

# ========== STEP 3: SETUP PATHS ==========
print("\n" + "="*50)
print("SETTING UP PATHS")
print("="*50)

# Paths based on your dataset
TRAIN_DIR = '/content/dataset_processing_output/augmented/classification/train'
VAL_DIR = '/content/dataset_processing_output/augmented/classification/val' if 'val' in items else TRAIN_DIR
TEST_DIR = '/content/dataset_processing_output/augmented/classification/test' if 'test' in items else VAL_DIR

print(f"Train: {TRAIN_DIR}")

print(f"Val: {VAL_DIR}")
print(f"Test: {TEST_DIR}")

# ========== STEP 4: LOAD DATA ==========
print("\n📥 Loading data...")

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Data generators
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    horizontal_flip=True
)

val_test_datagen = ImageDataGenerator(rescale=1./255)

# Load training data
train_gen = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical'
)

# Load validation data
val_gen = val_test_datagen.flow_from_directory(
    VAL_DIR,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)

print(f"\n✅ Data loaded!")
print(f"Training images: {train_gen.samples}")
print(f"Classes: {list(train_gen.class_indices.keys())}")

# ========== STEP 5: BUILD & TRAIN MODELS ==========
print("\n" + "="*50)
print("TRAINING MODELS")
print("="*50)

from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16, ResNet50, MobileNetV2, EfficientNetB0

def build_and_train(model_name, base_model_class):
    """Build and train a single model"""
    print(f"\n🚀 Training {model_name}...")

    # Build model
    base = base_model_class(include_top=False, weights='imagenet', input_shape=(224, 224, 3))
    base.trainable = False

    model = models.Sequential([
        base,
        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(train_gen.num_classes, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    # Train
    history = model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=15,
        verbose=1
    )

    # Evaluate
    val_loss, val_acc = model.evaluate(val_gen, verbose=0)
    print(f"✅ {model_name} Accuracy: {val_acc:.4f}")

    # Save model
    model.save(f'/content/{model_name}_trained.h5')

    return val_acc

# Train all 4 models
models_to_train = [
    ('VGG16', VGG16),
    ('ResNet50', ResNet50),
    ('MobileNetV2', MobileNetV2),
    ('EfficientNetB0', EfficientNetB0)
]

results = {}
for name, model_class in models_to_train:
    try:
        acc = build_and_train(name, model_class)
        results[name] = acc
    except Exception as e:
        print(f"❌ Error with {name}: {e}")

# ========== STEP 6: SHOW RESULTS ==========
if results:
    print("\n" + "="*50)
    print("RESULTS SUMMARY")
    print("="*50)

    import pandas as pd

    # Create results table
    df = pd.DataFrame({
        'Model': list(results.keys()),
        'Accuracy': list(results.values())
    }).sort_values('Accuracy', ascending=False)

    print("\n📊 Model Performance:")
    print(df.to_string(index=False))

    # Find best model
    best_model = df.iloc[0]
    print(f"\n🏆 Best Model: {best_model['Model']} ({best_model['Accuracy']:.4f})")

    # Save results
    df.to_csv('/content/model_results.csv', index=False)

    # Plot comparison
    import matplotlib.pyplot as plt

    plt.figure(figsize=(10, 6))
    bars = plt.bar(df['Model'], df['Accuracy'], color=['blue', 'green', 'orange', 'red'])

    for bar, acc in zip(bars, df['Accuracy']):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
                f'{acc:.4f}', ha='center', va='bottom')

    plt.title('Model Accuracy Comparison', fontsize=14, fontweight='bold')
    plt.xlabel('Model')
    plt.ylabel('Accuracy')
    plt.ylim(0, 1)
    plt.grid(True, alpha=0.3, axis='y')
    plt.tight_layout()
    plt.savefig('/content/model_comparison.png', dpi=100)
    plt.show()
else:
    print("❌ No models were trained successfully")

# ========== STEP 7: DOWNLOAD MODELS ==========
print("\n📁 Your trained models:")
!ls -la /content/*.h5

print("\n📊 Results files:")
!ls -la /content/*.csv /content/*.png 2>/dev/null || echo "No results files found"

# Download option
print("\n📥 To download models:")
print("1. Click the folder icon on the left (📁)")
print("2. Find your .h5 files")
print("3. Right-click and select 'Download'")
print("\nOr run this to create a zip:")
print("!zip -r /content/trained_models.zip /content/*.h5 /content/*.csv /content/*.png")

# Create zip
!zip -r /content/trained_models.zip /content/*.h5 /content/*.csv /content/*.png 2>/dev/null || echo "Creating zip of available files"

print("\n✅ All done! Models are saved in /content/")
print("📦 Zip file: /content/trained_models.zip")
```
